```python
import os
from ultralytics import RTDETR

PROJECT_ROOT = os.getcwd()
output_dir = os.path.join(PROJECT_ROOT, "3_models")
yaml_path = os.path.join(PROJECT_ROOT, "scripts", "finetune_blood_cell.yaml")

print("Starting High-Resolution Fine-Tuning (Target: 96%+ mAP)...")

# EXACT SAME pre-trained weights from your first stage
pretrained_weights = os.path.join(PROJECT_ROOT, "3_models", "stage1_pretrained", "weights", "best.pt")
model = RTDETR(pretrained_weights)

# New optimized run for dense cells
results = model.train(
    data=yaml_path,
    epochs=100,            
    patience=25,           
    imgsz=800,             # Increased from 640 for better small object detection
    batch=4,               # Decreased to 4 to save GPU VRAM
    workers=4,
    project=output_dir,
    name='stage2_finetuned_highres', # Saving to a new folder
    device=0,
    
    # Surgical Augmentations
    lr0=0.001,             
    lrf=0.01,
    degrees=90.0,          
    fliplr=0.5,            
    flipud=0.5,            
    mosaic=1.0,            
    mixup=0.0              # Disabled mixup so RBCs don't overlap artificially
)
print("High-Resolution Fine-tuning completed!")
```

```python
print("Generating Paper-Ready Validation & Test Metrics...")

highres_weights = os.path.join(PROJECT_ROOT, "3_models", "stage2_finetuned_highres", "weights", "best.pt")
best_model = RTDETR(highres_weights)

# 1. Internal Testing (The main benchmark)
print("Benchmarking on Internal Test Set...")
test_metrics = best_model.val(
    data=yaml_path,
    split='test',
    project=output_dir,
    name='stage2_eval_highres_test',
    save_json=True,       
    plots=True,
    conf=0.25,            # Slightly higher confidence to reduce False Positives in RBCs
    iou=0.60              # Better NMS handling for overlapping RBCs
)

print("High-Res Evaluation complete! Check '3_models/stage2_eval_highres_test' for results.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

results_path = os.path.join(PROJECT_ROOT, "3_models", "stage2_finetuned_highres", "results.csv")

if os.path.exists(results_path):
    df = pd.read_csv(results_path)
    df.columns = df.columns.str.strip()

    train_loss_col = [c for c in df.columns if 'train' in c and ('giou_loss' in c or 'box_loss' in c)][0]
    val_loss_col = [c for c in df.columns if 'val' in c and ('giou_loss' in c or 'box_loss' in c)][0]

    plt.figure(figsize=(10, 6))
    sns.set_theme(style="whitegrid")
    sns.lineplot(data=df, x='epoch', y=train_loss_col, label='Train Loss', linewidth=2, color='darkblue')
    sns.lineplot(data=df, x='epoch', y=val_loss_col, label='Validation Loss', linewidth=2, color='darkorange')
    
    plt.title('Training vs Validation Loss (High-Res 800px)', fontsize=14, fontweight='bold')
    plt.xlabel('Epochs', fontsize=12)
    plt.ylabel('Loss Value', fontsize=12)
    plt.legend()
    sns.despine()
    
    plot_path = os.path.join(PROJECT_ROOT, "3_models", "stage2_finetuned_highres", "highres_train_val_curve.png")
    plt.savefig(plot_path, dpi=300, bbox_inches='tight')
    plt.show()
    print("New High-Res Loss curve generated!")
```

```python
import os
import shutil
from ultralytics import YOLO

PROJECT_ROOT = os.getcwd()
output_dir = os.path.join(PROJECT_ROOT, "3_models")
train_dir = os.path.join(output_dir, "yolo11_stage2_finetune") # Loss curve er folder
eval_dir = os.path.join(output_dir, "yolo11_eval_internal_test") # mAP curves er folder
test_images_dir = os.path.join(PROJECT_ROOT, "2_preprocessed_datasets", "bccd_finetune", "images", "test")
best_weights = os.path.join(train_dir, "weights", "best.pt")

# Paper-er jonno ekta dedicated folder
paper_dir = os.path.join(output_dir, "paper_visuals")
os.makedirs(paper_dir, exist_ok=True)

print("Preparing High-Quality Visuals for the Research Paper...")

# --- Step 1: Extract Core Graphs (From respective folders) ---
print("Extracting evaluation and training graphs...")

# 1. Training Graphs (from stage2_finetune)
train_graphs = ['results.png']
# 2. Validation Graphs (from eval_internal_test)
eval_graphs = ['confusion_matrix_normalized.png', 'PR_curve.png', 'F1_curve.png']

graphs_found = 0

# Copy Training Graphs
for graph in train_graphs:
    src_path = os.path.join(train_dir, graph)
    dest_path = os.path.join(paper_dir, f"training_loss_{graph}") # Renamed for clarity
    if os.path.exists(src_path):
        shutil.copy(src_path, dest_path)
        graphs_found += 1
    else:
        print(f"Warning: Could not find {graph} in {train_dir}")

# Copy Validation Graphs
for graph in eval_graphs:
    src_path = os.path.join(eval_dir, graph)
    dest_path = os.path.join(paper_dir, graph)
    if os.path.exists(src_path):
        shutil.copy(src_path, dest_path)
        graphs_found += 1
    else:
        print(f"Warning: Could not find {graph} in {eval_dir}")

print(f"Successfully copied {graphs_found}/4 paper-ready graphs to 'paper_visuals'.")

# --- Step 2: Generate "Visually Stunning" Inference Images ---
print("Generating crisp bounding box predictions for the paper...")

model = YOLO(best_weights)

try:
    results = model.predict(
        source=test_images_dir,
        project=paper_dir,
        name='stunning_predictions',
        conf=0.45,            
        iou=0.50,             
        save=True,            
        line_width=2,         
        show_labels=True,     
        show_conf=False       
    )
    print("High-quality prediction images saved in 'paper_visuals/stunning_predictions'.")
except Exception as e:
    print(f"Error generating images: {e}")

print("\nAll visuals are ready! Please check the '3_models/paper_visuals' folder.")
```
